##Find vector W

$$
\begin{bmatrix}
w_1 \\
w_2 \\
\vdots \\
w_n
\end{bmatrix}
\times
\begin{bmatrix}
x_1^{(1)} & x_2^{(1)} & \cdots & x_n^{(1)} \\
x_1^{(2)} & x_2^{(2)} & \cdots & x_n^{(2)} \\
x_1^{(3)} & x_2^{(3)} & \cdots & x_n^{(3)} \\
\vdots & \vdots & \ddots & \vdots \\
x_1^{(m)} & x_2^{(m)} & \cdots & x_n^{(m)}
\end{bmatrix}
+ b =
\begin{bmatrix}
y^{(1)} & y^{(2)} & \dots & y^{(m)}
\end{bmatrix}
$$

or it can be interpreted as:
$$
W.X + b = Y
$$
to find W, we have:

$$
W= \frac{\text{Y  -  b}}{\text{X}}\huge
$$
given that Y-b as b has been accounted inside y
$$
\begin{bmatrix}
y^{(1)}-b & y^{(2)}-b & \dots & y^{(m)}-b
\end{bmatrix}
$$
with X^T (or in python we can make this as `X.T`), X^T is the transpose of X, with the column -> row and vice versa, meaning the X is in n x m size, the X^T is m x n

$$
X^T(X.W - Y) = 0
$$
$$
X^T.X.W - X^T.Y = 0
$$
$$
X^T.X.W = X^T.Y
$$
$$
w=(X^T.X)^{−1}X^T.y
$$

So to find w, except using W = (y-b)/X , we can use that to find the vector w


Given an model, derives from this predisposed model -- a linear regression one

$$
y=3x_1 - 5x_2 +6x_3 + 3
$$

For the second class StandardScalar standardizes features by removing the mean and scaling to unit variance. It transforms each features so that it has a mean of 0 and a standard dev = 1. This is also called z-score norm
***Why use it?***

- Many machine learning algorithms (e.g., SVM, neural networks, gradient descent-based models) perform better when features are on the same scale.

- It ensures that features with larger ranges don’t dominate those with smaller ranges.

- It’s critical when features have different units or scales (e.g., height in cm vs. weight in kg).
- How StandardScalar computes:

$$
z= \frac{\text{x - μ}}{\text{σ}}\huge
$$
with x: mean with any possible values

σ: Standard Deviation of the feature

x: Standardized features (mean=0, std+1)

*Key Features*

`fit(X)`: Computes the mean and standard deviation for each feature in X.

`transform(X)`: Applies the standardization using the computed mean and std.

`inverse_transform (X_scaled)`: Reverses the standardization to get back original values.

`fit_transform(X)`: Combines fit and transform in one step.

***make_pipeline*** is a utility function in scikit-learn’s pipeline module that creates a Pipeline object to sequentially apply a series of transformations (e.g., preprocessing steps like P`olynomialFeatures and StandardScaler`) and, optionally, a final estimator (e.g., a model like `LinearRegression`) to input data. It ***simplifies the process of chaining multiple steps into a single***

***Why use it?***

***Streamlines Workflow***: Combines multiple preprocessing steps and a model into one object, ensuring they’re applied in the correct order.

***Prevents Data Leakage***: Ensures transformations (e.g., scaling) are fit only on training data, not test data.

***Simplifies Code***: Reduces boilerplate code compared to manually applying each step.

***Reusability***: The pipeline can be treated as a single estimator, making it easy to fit, predict, or cross-validate.

***How it Works***

`make_pipeline `takes a sequence of estimators (e.g., transformers like `PolynomialFeatures, StandardScaler`, or models like `LinearRegression`) and creates a Pipeline object. Each step is applied in order:

`Transformers` (e.g., `PolynomialFeatures, StandardScaler`) process the input data using fit_transform during training.

The final estimator (if included) is trained on the transformed data using `fit`.

For prediction, the pipeline applies `transform` (for transformers) and `predict` (for the final estimator).

`mean_squared_error (MSE)` is a function in scikit-learn that calculates the mean squared error, a common metric to measure the error (or performance) of a regression model. It quantifies how far off your model’s predictions are from the actual values by averaging the squared differences between predicted and true values.

Why use it?

***Evaluates Regression Models:*** MSE measures how well a regression model (e.g., linear regression) predicts continuous values.

***Penalizes Large Errors***: Since errors are squared, larger deviations contribute more to the final score, making it sensitive to outliers.

***Interpretable***: Lower MSE indicates better model performance (closer predictions to actual values).

***Common Standard***: Widely used in machine learning to compare models or tune hyperparameters.

MSE: MEAN SQUARED ERROR
$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$
yi: actual value for the i-th sample

y^i: Predicted value for the i-th sample


**Key Parameters**

When calling `mean_squared_error(y_true, y_pred, **kwargs):`

`y_true`: Array of actual target values.

`y_pred`: Array of predicted values.

`squared`: If True (default), `returns MSE`; if False, returns the root mean squared error (RMSE).

`sample_weight`: Optional weights for each sample (e.g., to emphasize certain data points).

`multioutput`: For multi-output tasks (e.g., predicting multiple variables), options are `'raw_values' `(MSE for each output) or `'uniform_average'` (average across outputs, default).

##Underfitting/Overfitting Behaviors and Bias/Variable tradeoffs

Underfitting: informally speaking, is when the model is too simple, the hypothesis about data distribution is wrong and too simple, this situation is also called high bias, means that your algorithms can do accurate predictions, but the initial assumption about the data is incorrect

Overfitting: informally speaking, is when the model is overly complex for the given data, the data distribution is wrong and basically too complex, the situation is called high variance. Means that your algorithms cannot do accurate predictions -- changing the input data only a little, but the model output changes very much

Bias/variance trade-off
 - low-bias, low-variance -- is a good result
 - low bias, high variance -- overfitting -- the algorithm output very different predictons for similar data
 - high bias, low variance -- underfitting -- the algorihm outputs similar predictions for similar data, but predictions are wrong (algorithm miss)
 - high bias, high variance-- very bad algorithm

````python
def polynomial_Ridge_regression(degree=1, alpha=1.0):

    return make_pipeline(PolynomialFeatures(degree),
                         StandardScaler(),
                         Ridge(alpha))
'''
just like before, but this emphasis on the alteration of the shape of the vector

the degree meaning the greatest zero the Feature is, like the largest x^n

the linear_model Ridge is actually a technique in ML used to address overfitting in
linear models, particularly dealing with high correlation between independent variables
which based on alpha.
'''
````

Ridge regression is useful when ***the goal is to minimize the impact of less important features while keeping all variables in the model***. Lasso regression is preferred when the goal is feature selection, resulting in a simpler and more interpretable model with fewer variables.

##Certain Mathematical function to validate the closeness of the data

Cách 1:
1. MSE - Mean Square Error
$$
\begin{align}
MSE(w) = \frac{1}{M}\sum_{i=1}^{M}(\hat{y}^i-y^i)^2
\end{align}
$$
2. RMSE - Root Mean Squared Error
$$
\begin{align}
RMSE(w) = \sqrt{MSE(w)} = \sqrt{\frac{1}{M}\sum_{i=1}^{M}(\hat{y}^i-y^i)^2)}
\end{align}
$$
3. MAE - Median Absoulute Error
$$
\begin{align}
MAE(w) = \frac{1}{M}\sum_{i=1}^{M}|\hat{y}^i-y^i|
\end{align}
$$
4. R-square => Coefficient of determination, the closer to 1 the better
$$
\begin{align}
R^2(w) = 1 - \frac{SS_{res}}{SS_{tot}} = 1 - \frac{\sum_i (y_i - \hat{y}_i)^2}{\sum_i (y_i - \bar{y})^2}
\end{align}
$$

````python
from sklearn.metrics import mean_squared_error, median_absolute_error, r2_score
import math
def getScores(y_predict, y_test):
    mse = mean_squared_error(y_predict, y_test)
    rmse = math.sqrt(mse)
    mae = median_absolute_error(y_predict,y_test)
    r2 = r2_score(y_predict, y_test)
    return {"MSE": mse, "RMSE": rmse, "MAE": mae, "R2_Score": r2 }
````

# 4.Gradient Descent

- In machine learning, pridicting and giving a functions from a given data though in this practice seems easy enough, in real world utilization, it is extrememly complex, regarding there can be various types of functions, polynomial, liear... let alone given that these fuctions can exists in a higher dimentions, not fixated on a 2D examples given before.


- Thus, finding an derivative equations is again extremely difficult, given the complexity of the derivative equation or the high-dimension data points.

- ***Gradient Descent*** solved the problem and become the most important part in lost of machine learning algorithms and deep learning models until now.


image.png

So what is the principle of Gradient Descent

Given a certain graphs, and it should obviously have a point--local minimum-- that has the value(y) as the local minimum or global minimum which is basically the smallest value in the graph

image.png

For instance, in this parabol, the smallest values is located at $x^*$, and the whole function is $f(x) = 1/2(x-1)^2 -2$

in math, to find that place, we should use the derivate to find it, and from this example. the derivative is: $y' = x-1$

Thus, we create a scienario, of x(s) locations. Assume if we let x directly into the minimum point, than there will be no arguments. But if it is not the case, it will be 2 different scinerios, one is $x_1$ is less than $x^*$, another is $x_2$ is larger than $x^*$

So basically, the principle is decreasing or increase the value x toward the $x^*$, and there is a certain functions for that

$$
x_{t+1}=xt−ηf′(xt)
$$

##Mathematically Proving the function

so as we taken from the preceding example $f'(x) = x-1$ and from the 2 scinarios:

if $f'(x)>0$, so $x_t > x^*$, and if $f'(x)<0$ then $x_t<x^*$

so, for any $x_t$, whether less than or more than $x^*$, with its next x's that is mathematically given to be closer than $x^*$ would have this equation:
$$x_{t+1}=x_t + δ$$
and the size of δ varies, it can be (-) if the $x_t$ is more than $x^*$ and vice versa, (+) if $x_t$ is less than $x^*$

in fact, the longer the distance between $x^*$ and $x_t$, the higher the value $|f'(x)|$, thus $δ=−ηf ′(x_t)$, with $f'(x)$ dictates the symbols, and the η is the learning pace, the pace for $x_t$ to kind of scans through the whole graph to find the minimum, making the final equations to be simply:

$$
x_{t+1}=xt−ηf′(xt)
$$

##Mathematical Theoretical Implementation in Python

given a certain 2D for instance.
$$f(x) = x^2 + 5sin(x)$$
$$f'(x) = 2x + 5cos(x)$$
$$x_{x+1} = x_t -η(2x+5cos(x))$$


image.png

with f(x) in purple, f'(x) in black

1dimg_5_0.1_5 (1).gif1dimg_5_0.1_-5.gif

And these are the principles

````python
import numpy as np
def function(x):
  return x ** 2 + 5 * np.sin(x)
def derivated_fuction(x):
  return 2 * x + 5 * np.cos(x)
````

````python
def gradient_descent_examples(learning_rate, x_0):
  x = [x_0]
  for _ in range(100):
    x_new = x[-1] - learning_rate*derivated_fuction(x[-1])
    #this is basically give a random ass x_0 on a graph, and then after this x_new
    #we have another x that have a distance between x* less than x_0 do. then we will do
    #ts continously until it reaches the designated point

    if abs(derivated_fuction(x_new))<1*10**-4: #the point to is make the derivate function to near 0 in order to get the smallest function value
      break

    x.append(x_new) #to continue decrease/increase the previous x_new
  return x
````

##Gradient Descent with Momentum.

given a initiate point of $x$

1. Basic Gradient Descend
$$
x_{t+1} = x_t - nf'(x_t)
$$


2. Gradient Descent With Momentum,
 given the momentum formula:
with $v_k$ stands for velocity
$$
v_{t+1} = γvt+ηf'(xt)
$$

with $γ$ as momentum rate, like the learning rate but it only depicts the rate for momentum.

and to insert into the momentum formula, we got the whole gradient descent with momentum as:

$$
x_{t+1} = x_t - v_{t+1}
$$


<b>disclaimer, the $v_{t+1}$ which is the velocity of the gradient descent function momentum should be updated per iterations. Just like in physics, there should be accelation for the incremation/decreasing of the object's velocity.</b>

momentum1d.gifnomomentum1d.gif


Example:

$$
 y=sin(x)+0.1x^2
$$

$$
y′=cos(x)+0.2x
$$

````python
import numpy as np

def function(x):
  return np.sin(x) + 0.1*x**2

def gradient(x):
  return np.cos(x) + 0.2*x

def testing_gradient_descent(x_initeration, learning_rate=0.1, iteration_time=1000):
  x = x_initeration
  for _ in range(iteration_time):
    x = x- learning_rate*gradient(x)
    if abs(gradient(x)) < 1*10**-6: break
  return x

print(testing_gradient_descent(-7))
print(testing_gradient_descent(7))
````

image.png

````python
def testing_gradient_descent_with_momentum(x_init, learning_rate=0.1, momentum_value=0.9,
                                           iteration_time=1000):
  x = x_init
  v = 0
  for _ in range(iteration_time):
    v = momentum_value * v + learning_rate * gradient(x)
    x = x - v
  return x
print(testing_gradient_descent_with_momentum(-8))
print(testing_gradient_descent_with_momentum(8))
````

## Nesterov accelerated gradient (NAG)

the Nesterov accelerated gradient is a optimization methods that is especially used in gradient descent

and these are the steps taken:

**1. predicts the next values**
at this step, we find the gradient at the location to execute $x_{ahead}$, this predicts by going back from the aprevious location = the function of the product between $η$ and and the gradient function.
$$x_{ahead}=x-ηf′(x_t)$$

**2.Find gradient and updating the velocity**
after locating the $x_{ahead}$, we find the gradient $f'(x_{ahead})$ of the loss function/function in genera. And then, we update the velocity vector by combining the gradient with the current velocity vector based on the learning speed and the momentum rate

$$v = γvt-ηf′(x_{ahead})$$

**3. updating the current value**
basically this, but instead of minus velocity, addition is use to supplanted.
$$x=x_{ahead}+v$$

image.png

````python
def gradient_descent_with_NAG(x_initiate, momentum_rate=0.9, learning_rate = 0.1, iteration_time =1000):
  x=x_initiate
  velocity = 0
  x_ahead = 0
  for _ in range(iteration_time):
    x_ahead = x - momentum_rate*velocity
    velocity = momentum_rate*velocity - learning_rate*gradient(x_ahead)
    x = x + velocity
  return x

print(gradient_descent_with_NAG(-7))
````

## Types of Gradient Descent

<h2>Batch Gradient Descent</h2>

also mean vanilla gradient descent, which ***calculates the error for each example withing the training datasets***, but only after all training examples have been evaluated does the model will then get upgraded since they have to recognize the trends that the dataset represents. the whole process ***is an cycle*** and is called an `***training epoch***`

Or put it simply, it computes the ***entire training dataset for each iteration***, while ensuring pricision, but can be computationally expensive when dealing with large data sets.

Advantages:
 1. Accurate Gradient Estimates.
 2. Good for Smooth Error Surface.

Disadvantages
 1. Slow Convergence: need a long time to converge
 2. High Memory Usage
 3. Insufficient for Large Datasets


<h2>Stochatic Gradient Descent(SGD)</h2>
This is a method to kind of address the inefficiency of Batch Gradient Descent by computing a gadient using a single training example, which just like linear method of splitting the dataset ideally 80/20 using




```
from sklearn.model_selection import train_test_split
```
This makes the algorithm much faster and have an excess amount of data for efficency testing

Advantages
 1. Faster ConvergenceL

##5.1 Rescaling/Min-max normalization

Resclaing is the simplest method and consists in rescaling the range of features to scale the range in [0,1] or [-1,1]. Selecting the target range depends on the nature of the data. the general formula for a [min,max] of [0,1]:

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$


with x is the normal value and x' is the normalized value

or to escalate it into a [a,b] value, this formula become:

$$x' = a+ \frac{(x - x_{min})(b-a)}{x_{max} - x_{min}}$$


````python
def min_max_scaling(df, a=0 , b=1):
    # copy the dataframe
    df_norm = df.copy()
    # apply min-max scaling
    for column in df_norm.columns:
        df_norm[column] = a+ (df_norm[column] - df_norm[column].min())*(b-a) / (df_norm[column].max() - df_norm[column].min())

    return df_norm
  #take x of the previous test
min_max_scaling(x,-1,1)
````

##5.2 Mean normalization

$$x' = \frac{x - \bar{x}}{x_{max} - x_{min}}$$

with $\bar{x} = a_{average}$, this is basically a way of feature scaling more than the Rescaling, Rescaling is actually just a way to minimize the range of the whole data base. While this is the way to do the normalization and decrease the fluctuation/smoothing methods. like the previous StandardScaler()

````python
def mean_normalization(df):
    # copy the dataframe
    df_norm = df.copy()
    # apply min-max scaling
    for column in df_norm.columns:
        df_norm[column] = (df_norm[column] - df_norm[column].mean()) / (df_norm[column].max() - df_norm[column].min())

    return df_norm

mean_normalization(x)
````

##5.3 Z-score normalization

after the integration of z-score normalization, all features will have a man of 0 and standard deviation of 1 with this mathematical formula:

$$x_j^{(i)} = \frac{x_j^{(i)} - μ_j}{σ_j}$$
or just
$$x = \frac{x - μ}{σ}$$

with μ as the mean and σ as the standard deviation

$$
\mu_j = \frac{1}{m} \sum_{i=0}^{m-1} x_j^{(i)}
$$

$$
\sigma_j^2 = \frac{1}{m} \sum_{i=0}^{m-1} \left(x_j^{(i)} - \mu_j\right)^2
$$

````python
# apply the z-score method in Pandas using the .mean() and .std() methods
def z_score(df):
    # copy the dataframe
    df_std = df.copy()
    # apply the z-score method
    for column in df_std.columns:
        df_std[column] = (df_std[column] - df_std[column].mean()) / df_std[column].std()

    return df_std
````
